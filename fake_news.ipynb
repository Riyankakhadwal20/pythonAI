```python
import pandas as pd

# Count Vectorizer:-
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.model_selection import train_test_split
cv= CountVectorizer()
data = pd.read_csv('Fake.csv')
data.head()
data.dropna(inplace=True)
data.loc[0,'title']
count= CountVectorizer()
count.fit(data.loc[:,'title'])
count.vocabulary_
len(count.vocabulary_)
a= count.transform(data.loc[:,'title'])
a=a.toarray()
a
data.loc[:,'vectors']= list(a)
data


# TF-IDF Vectorizer:-
from sklearn.feature_extraction.text import TfidfVectorizer
tf= TfidfVectorizer()
data = pd.read_csv('Fake.csv')
data.head()
data.dropna(inplace=True)
tf.fit(data.loc[:,'subject'])
tf.vocabulary_
len(tf.vocabulary_)
a=tf.transform(data.loc[:,'subject']).toarray()
len(a[0])
a[0]
f= tf.get_feature_names_out()
f
for i in f:
    index = tf.vocabulary_.get(i)
    print(i,tf.idf_[index])
```
