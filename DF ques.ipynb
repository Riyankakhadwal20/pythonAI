```python
import pandas as pd

# 1. To download NEET Tweets dataset
df = pd.read_csv("neet_data.csv")
df.head()

# 2. Import CSV and Show Column information and total rows and column
df.info()
print("Total columns:", len(df.columns))
print("Total rows:", df.shape[0])

# 3. Sort the dataset with highest retweet to low
sorted_df = df.sort_values(by="retweet_count", ascending=False)
print(sorted_df.head())

# 4. Show the tweet source, user_name, tweet, created_at and user_url highest value of retweet count 
max_retweet = df.loc[df['retweet_count'].idxmax()]
result = max_retweet[['source', 'retweet_count', 'user_name', 'tweet', 'created_at', 'user_url']]
print(result)

# 5. highest top 10 retweet count value print along with text heading like "TOP 10 Retweet sum is XXXXXX"
top10 = df.sort_values(by="retweet_count", ascending=False).head(10)
total_retweets = top10["retweet_count"].sum()
print("TOP 10 Retweet sum is", total_retweets)
print("\nTop 10 Tweets by Retweet Count:")
print(top10[["user_name", "tweet", "retweet_count"]])

# 6. Count and show Total number of retweet done....
android_tweets = df[df["source"] == "Twitter for Android"]
count_android = android_tweets.shape[0]
sum_android = android_tweets["retweet_count"].sum()
print("Total retweet via Twitter for Android:", count_android)
print("Total retweet sum via Twitter for Android:", sum_android)

# 7. Print the username and location of person having highest follower count.
max_followers = df.loc[df["user_follower_count"].idxmax()]
print("Username:", max_followers["user_name"])
print("Location:", max_followers["user_location"])

# 8. Create a new column with name followers_friend_ratio....
df["followers_friend_ratio"] = df["user_follower_count"] / df["user_friends_count"]
print(df[["user_follower_count", "user_friends_count", "followers_friend_ratio"]].head())

# 9. Tell total null values count of each column
null_counts = df.isnull().sum()
print("Total null values in each column:",null_counts)

# 10. Fill user_description as Twitter user if it is NA value on original DF
df["user_description"].fillna("Twitter user", inplace=True)
print(df["user_description"].head())

# 11. Drop all columns if they have any null value and change original DF
print("Before Dropping Columns:",df)
df.dropna(axis=1, inplace=True)
print("\nAfter Dropping Columns with Null Values:",df)
print(df)

# 12. Create a new column at 3rd location with name is_allow as True. in original DF
df.insert(2, "is_allow", True)
print(df.head())

# 13. Drop the column tweet and print final DF
df.drop("tweet", axis=1, inplace=True)
print("Final DataFrame:\n")
print(df)

# 14. Convert created_at into string and print the month and year. 
df["created_at"] = df["created_at"].astype(str)
split_date = df["created_at"].str.split("-")
year = split_date.str[0]
month = split_date.str[1]
print("Year:",year.head())
print("\nMonth:",month.head())
```
