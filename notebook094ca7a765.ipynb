```python
import os
import tensorflow as tf 
import matplotlib.pyplot as plt
from tensorflow.keras.models import Sequential, Model 
from tensorflow.keras.layers import Dense, Flatten, Dropout, MaxPooling2D
```

```python
from tensorflow.keras.preprocessing.image import ImageDataGenerator
```

```python
train_datagen= ImageDataGenerator(rescale=1./255, zoom_range=0.2)  
val_datagen= ImageDataGenerator(rescale=1./255)
```

```python
train_generator = train_datagen.flow_from_directory(
    '/kaggle/input/datasets/mukaffimoin/potato-diseases-datasets',
    target_size=(224,224),  # 
    batch_size=16,  # Instead of sending all images at once, it sends 32 images at a time.
    class_mode='categorical'
)
```

```python
classes =train_generator.class_indices  
classes
```

```python
classes = {j:i for i,j in classes.items()}
classes
```

```python
val_generator = val_datagen.flow_from_directory(
    '/kaggle/input/datasets/mukaffimoin/potato-diseases-datasets',
    target_size=(224,224),
    batch_size=16,
    class_mode='categorical'
)
```

```python
from tensorflow.keras import datasets, layers, models
```

```python
model = models.Sequential([
    layers.Conv2D(filters=32, kernel_size=(3,3),activation='relu',input_shape=(224,224,3)),
    layers.MaxPooling2D((2,2)), 

    layers.Conv2D(filters=64, kernel_size=(3,3),activation='relu'),
    layers.MaxPooling2D((2,2)),
    
    layers.Conv2D(filters=128, kernel_size=(3,3),activation='relu'),
    layers.MaxPooling2D((2,2)),
    
    layers.Flatten(),
    layers.Dense(128, activation='relu'),
    layers.Dense(64, activation='relu'),
    layers.Dense(7, activation='softmax') 
])
```

```python
model.compile(optimizer='adam', loss='categorical_crossentropy', metrics=['accuracy'])
```

```python
model.fit(train_generator,batch_size=16, epochs=20, validation_data= val_generator)
```

```python
model.save("potato.keras")
```
