```python
import pandas as pd

# 1. How would you access the value corresponding to the index 'y' using .loc[]?
data = pd.Series([100,200,300,400],index = ['x','y','z','w'])
value = data.loc['y']        
print(value)       

# 2. How would you access the 2nd element in the series using .iloc[]?
second_element = data.iloc[1]
print(second_element)

# 3. How would you access multiple elements, say for indices 'x' and 'w', using .loc[]?
multiple_elements = data.loc[['x','w']]
print(multiple_elements)

# 4. With the same Series, how can you access the last element using .iloc[]?
last_element = data.iloc[-1]
print(last_element)

# 5. How would you access elements from index 'y' to 'z' (inclusive) using .loc[]?
range_elements = data.loc['y':'z']
print(range_elements)

# 6. What is the difference between using .iloc[2] and .loc[2] when accessing a Series that has integer-based indices?
data = pd.Series([100, 200, 300, 400], index=[0, 1, 2, 3])
data.iloc[2]    # position-based
data.loc[2]     # label-based

# 7. Using dot notation, how would you access the element with index 'x'?
element_x = data.x
print(element_x)
```
